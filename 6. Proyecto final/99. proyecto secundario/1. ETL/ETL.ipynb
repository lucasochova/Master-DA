```python
import numpy as np
import pandas as pd
```

### Importamos DATOS de cada año

```python

Datos_2015 = pd.read_csv("2015.csv")
Datos_2016 = pd.read_csv("2016.csv")
Datos_2017 = pd.read_csv("2017.csv")
Datos_2018 = pd.read_csv("2018.csv")
Datos_2019 = pd.read_csv("2019.csv")

pd.options.display.max_columns=30
Datos_2015.head(3)
```

### Agregamos culumna AÑO para poder identificar los valores despues de la fusion.

```python
Datos_2015["AÑO"] = 2015
Datos_2016["AÑO"] = 2016
Datos_2017["AÑO"] = 2017
Datos_2018["AÑO"] = 2018
Datos_2019["AÑO"] = 2019
```

### Revisamos las columnas de cada DF para hacer coincidir los nombres y tener datos homogeneos

```python
for anio, df in zip(range(2015, 2020), [Datos_2015, Datos_2016, Datos_2017, Datos_2018, Datos_2019]):
    print(anio, list(df.columns))
```

### Unificamos terminos y unificamos los DF

```python
mapa = {
    # Rank
    "Happiness Rank": "Rank",
    "Happiness.Rank": "Rank",
    "Overall rank": "Rank",
    # País
    "Country or region": "Country",
    # Score
    "Happiness Score": "Score",
    "Happiness.Score": "Score",
    # PIB
    "Economy (GDP per Capita)": "GDP",
    "Economy..GDP.per.Capita.": "GDP",
    "GDP per capita": "GDP",
    # Apoyo social
    "Family": "Social support",
    # Esperanza de vida
    "Health (Life Expectancy)": "Life expectancy",
    "Health..Life.Expectancy.": "Life expectancy",
    "Healthy life expectancy": "Life expectancy",
    # Libertad
    "Freedom to make life choices": "Freedom",
    # Corrupción
    "Trust (Government Corruption)": "Corruption",
    "Trust..Government.Corruption.": "Corruption",
    "Perceptions of corruption": "Corruption",
    # Residual de distopía
    "Dystopia Residual": "Dystopia Residual",
    "Dystopia.Residual": "Dystopia Residual",
}

lista = [Datos_2015, Datos_2016, Datos_2017, Datos_2018, Datos_2019]

dfs = []
for anio, df in zip(range(2015, 2020), lista):
    df = df.rename(columns=mapa)
    df["AÑO"] = anio
    dfs.append(df)


# solo columnas comunes a los 5 años
Datos_comunes = pd.concat(dfs, ignore_index=True, join="inner")
```

### Verificamos que todos los datos esten correctos, y que no haya nulls

```python
print(Datos_comunes.shape)
print(Datos_comunes.groupby("AÑO").size())
print(Datos_comunes.isna().sum())
```

### Exportamos los datos procesados para seguir trabajando en otra plataforma

```python
Datos_comunes.to_csv(
    'Datos_Limpios15-19.csv',
    index=False,          # no incluir el índice del DataFrame
    sep=',',              
    encoding='utf-8-sig', 
    header=True           # incluir (o no) los nombres de columna
)
```
